```python
#def prime factor
def prime_factors(n):
    factors = []
    i = 2

    while i <= n:
        if n % i == 0:
            factors.append(i)
            n = n // i
        else:
            i += 1

    return factors


num = int(input("Enter a number: "))
print("Prime factors:", prime_factors(num))
```
