```python
# greatest of 3 no's 
a= float(input("Enter the first number: "))
b = float(input("Enter the second number: "))
c = float(input("Enter the third number: "))


if a>b>c:
    print("The greatest no is a",a)
 
elif b>c>a:
    print("The greatest number is",b)
    
else:
    print("The greatest number is",c)
```
