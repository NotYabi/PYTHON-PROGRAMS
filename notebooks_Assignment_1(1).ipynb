```python
# 1. Creating different data structures for students

students_list = ["YABI", "ADI", "AAYUSH"]

students_tuple = ("YABI", "ADI", "AAYUSH")

students_dict = {
    101: "YABI",
    102: "ADI",
    103: "AAYUSH"
}

print("--- Initial Data Structures")
print(f"List:       {students_list}")
print(f"Tuple:      {students_tuple}")
print(f"Dictionary: {students_dict}\n")


# 2. Performing Dictionary Operations

# --- OPERATION 1: ADD ---
students_dict[104] = "David"
print("--- After Adding Roll 104 (David) ---")
print(f"Dictionary: {students_dict}\n")


# --- OPERATION 2: UPDATE ---
students_dict[102] = "Robert"
print("--- After Updating Roll 102 to 'Robert' ---")
print(f"Dictionary: {students_dict}\n")


# --- OPERATION 3: DELETE ---
if 103 in students_dict:
    del students_dict[103]
print("--- After Deleting Roll 103 ---")
print(f"Dictionary: {students_dict}\n")


# Final State summary
print(f"List (Unchanged):       {students_list}")
print(f"Tuple (Immutable):      {students_tuple}")
print(f"Dictionary (Modified):  {students_dict}")
```
